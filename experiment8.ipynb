```python
# ===========================================
# Case Study : Smart Agriculture Monitoring System
# Google Colab Simulation
# ===========================================

import random
import time
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import clear_output

# -------------------------
# Field Names
# -------------------------

fields = [
    "Field 1","Field 2","Field 3"
]

soil_data=[]
humidity_data=[]
temperature_data=[]

# -------------------------
# Simulation
# -------------------------

for field in fields:

    soil=random.randint(20,90)
    humidity=random.randint(40,90)
    temp=round(random.uniform(20,38),1)

    soil_data.append(soil)
    humidity_data.append(humidity)
    temperature_data.append(temp)

    if soil<35 or humidity<45 or temp>35:
        status="ALERT"
        color="red"
    else:
        status="NORMAL"
        color="green"

    clear_output(wait=True)

    print("="*60)
    print("SMART AGRICULTURE MONITORING SYSTEM")
    print("="*60)
    print("Field :",field)
    print("Soil Moisture :",soil,"%")
    print("Humidity :",humidity,"%")
    print("Temperature :",temp,"°C")
    print("Status :",status)

    plt.figure(figsize=(12,4))

    # Soil Moisture
    plt.subplot(1,3,1)
    plt.bar(["Soil"],[soil],color="brown")
    plt.ylim(0,100)
    plt.title("Soil Moisture")

    # Humidity
    plt.subplot(1,3,2)
    plt.bar(["Humidity"],[humidity],color="blue")
    plt.ylim(0,100)
    plt.title("Humidity")

    # Temperature
    plt.subplot(1,3,3)
    plt.bar(["Temp"],[temp],color=color)
    plt.ylim(15,45)
    plt.title("Temperature")

    plt.tight_layout()
    plt.show()

    time.sleep(2)

# -------------------------
# Final Report
# -------------------------

report=pd.DataFrame({

    "Field":fields,
    "Soil Moisture":soil_data,
    "Humidity":humidity_data,
    "Temperature":temperature_data

})

print("\nFINAL REPORT\n")
print(report)

report.to_csv("Smart_Agriculture_Report.csv",index=False)

print("\nCSV File Saved Successfully")
```
