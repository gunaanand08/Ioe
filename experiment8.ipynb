```python
# ===========================================
# Case Study : Healthcare Monitoring System
# Google Colab Simulation
# ===========================================

import random
import time
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import clear_output

# -------------------------
# Patient Names
# -------------------------

patients = [
    "Patient 1","Patient 2","Patient 3","Patient 4","Patient 5",
    "Patient 6","Patient 7","Patient 8","Patient 9","Patient 10"
]

temperature_data=[]
heart_data=[]
spo2_data=[]

# -------------------------
# Simulation
# -------------------------

for patient in patients:

    temp=round(random.uniform(35.5,40.5),1)
    heart=random.randint(55,130)
    spo2=random.randint(88,100)

    temperature_data.append(temp)
    heart_data.append(heart)
    spo2_data.append(spo2)

    if temp>38 or heart>110 or spo2<92:
        status="EMERGENCY"
        color="red"
    else:
        status="NORMAL"
        color="green"

    clear_output(wait=True)

    print("="*60)
    print("HEALTHCARE MONITORING SYSTEM")
    print("="*60)
    print("Patient :",patient)
    print("Temperature :",temp,"°C")
    print("Heart Rate :",heart,"BPM")
    print("SpO2 :",spo2,"%")
    print("Status :",status)

    plt.figure(figsize=(12,4))

    # Temperature
    plt.subplot(1,3,1)
    plt.bar(["Temp"],[temp],color="orange")
    plt.ylim(30,45)
    plt.title("Temperature")

    # Heart Rate
    plt.subplot(1,3,2)
    plt.bar(["Heart"],[heart],color="blue")
    plt.ylim(40,150)
    plt.title("Heart Rate")

    # SpO2
    plt.subplot(1,3,3)
    plt.bar(["SpO2"],[spo2],color=color)
    plt.ylim(80,100)
    plt.title("SpO2")

    plt.tight_layout()
    plt.show()

    time.sleep(2)

# -------------------------
# Final Report
# -------------------------

report=pd.DataFrame({

    "Patient":patients,
    "Temperature":temperature_data,
    "Heart Rate":heart_data,
    "SpO2":spo2_data

})

print("\nFINAL REPORT\n")
print(report)

report.to_csv("Healthcare_Report.csv",index=False)

print("\nCSV File Saved Successfully")
```
