```python
import random
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------------------
# Smart Forest Fire Detection System
# ----------------------------------------

num_sensors = 100

sensor_id = []
temperature = []
humidity = []
smoke_level = []
status = []

normal = []
warning = []
fire_risk = []

for i in range(1, num_sensors + 1):

    temp = round(random.uniform(24, 48), 1)      # °C
    hum = random.randint(20, 90)                 # %
    smoke = random.randint(50, 600)              # ppm

    # Forest Fire Detection Logic
    if temp >= 42 or smoke >= 450:
        s = "Fire Risk"
        fire_risk.append(i)

    elif temp >= 35 or smoke >= 250 or hum <= 35:
        s = "Warning"
        warning.append(i)

    else:
        s = "Normal"
        normal.append(i)

    sensor_id.append(i)
    temperature.append(temp)
    humidity.append(hum)
    smoke_level.append(smoke)
    status.append(s)

# ----------------------------------------
# Create DataFrame
# ----------------------------------------

df = pd.DataFrame({
    "Sensor ID": sensor_id,
    "Temperature (°C)": temperature,
    "Humidity (%)": humidity,
    "Smoke Level (ppm)": smoke_level,
    "Status": status
})

# Display Dataset
print(df)

# ----------------------------------------
# Summary
# ----------------------------------------

print("\n========== FOREST FIRE SUMMARY ==========")
print("Total Sensors      :", num_sensors)
print("Normal Areas       :", len(normal))
print("Warning Areas      :", len(warning))
print("Fire Risk Areas    :", len(fire_risk))

print("\nFire Risk Sensor IDs:")
print(fire_risk)

print("\nWarning Sensor IDs:")
print(warning)

if len(fire_risk) >= 15:
    print("\nForest Status : HIGH FIRE RISK")
elif len(fire_risk) >= 5:
    print("\nForest Status : MODERATE FIRE RISK")
else:
    print("\nForest Status : SAFE")

# ----------------------------------------
# Save Dataset
# ----------------------------------------

df.to_csv("forest_fire_data.csv", index=False)

# ----------------------------------------
# Visualization
# ----------------------------------------

plt.figure(figsize=(15,5))

colors = []

for s in status:
    if s == "Normal":
        colors.append("green")
    elif s == "Warning":
        colors.append("orange")
    else:
        colors.append("red")

plt.bar(sensor_id, temperature, color=colors)

plt.xlabel("Sensor ID")
plt.ylabel("Temperature (°C)")
plt.title("Smart Forest Fire Detection System")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```
