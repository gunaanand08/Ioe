```python
# ============================================
# Smart City Traffic Management System
# Dynamic Density-Based Traffic Light Controller
# Google Colab Version
# ============================================

import random
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time

# Lane names
lanes = ["North", "South", "East", "West"]

# Initial queue of cars
queues = [5, 2, 8, 3]

# Number of traffic cycles
simulation_steps = 20

for step in range(1, simulation_steps + 1):

    # Random new cars arrive
    new_cars = [random.randint(0, 5) for _ in range(4)]
    queues = [queues[i] + new_cars[i] for i in range(4)]

    # Find busiest lane
    active_lane = queues.index(max(queues))
    active_lane_name = lanes[active_lane]

    # Cars cleared
    cars_cleared = min(queues[active_lane], random.randint(4, 8))
    queues[active_lane] -= cars_cleared

    # Clear previous graph
    clear_output(wait=True)

    # Colors
    colors = ['red', 'red', 'red', 'red']
    colors[active_lane] = 'green'

    # Plot graph
    plt.figure(figsize=(8,5))
    plt.bar(lanes, queues, color=colors)

    plt.ylim(0,30)
    plt.ylabel("Number of Cars Waiting")
    plt.xlabel("Traffic Lanes")
    plt.title(f"Traffic Cycle {step}: Green Light -> {active_lane_name}")

    # Display values on bars
    for i, value in enumerate(queues):
        plt.text(i, value+0.5, str(value),
                 ha='center', fontsize=12)

    plt.figtext(
        0.5,
        0.01,
        f"Action: Cleared {cars_cleared} cars from {active_lane_name}",
        ha='center',
        fontsize=12,
        bbox=dict(facecolor='white', edgecolor='black')
    )

    plt.show()

    # Wait before next cycle
    time.sleep(1.5)

print("Simulation Completed Successfully!")
```
