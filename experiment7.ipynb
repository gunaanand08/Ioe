```python
# ============================================
# Smart City Water Management System
# Dynamic Level-Based Water Pump Controller
# Google Colab Version
# ============================================

import random
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time

# Tank names

tanks = ["North", "South", "East", "West"]

# Initial water level of tanks

levels = [45, 30, 65, 40]

# Number of monitoring cycles

simulation_steps = 20

for step in range(1, simulation_steps + 1):

    # Random water consumption
    water_used = [random.randint(0, 8) for _ in range(4)]
    levels = [max(0, levels[i] - water_used[i]) for i in range(4)]

    # Find tank with lowest water level
    active_tank = levels.index(min(levels))
    active_tank_name = tanks[active_tank]

    # Water supplied
    water_added = random.randint(5, 12)
    levels[active_tank] += water_added

    # Clear previous graph
    clear_output(wait=True)

    # Colors
    colors = ['blue', 'blue', 'blue', 'blue']
    colors[active_tank] = 'green'

    # Plot graph
    plt.figure(figsize=(8,5))
    plt.bar(tanks, levels, color=colors)

    plt.ylim(0,100)
    plt.ylabel("Water Level (%)")
    plt.xlabel("Storage Tanks")
    plt.title(f"Water Cycle {step}: Pump Activated -> {active_tank_name}")

    # Display values on bars
    for i, value in enumerate(levels):
        plt.text(i, value+1, str(value) + "%",
                 ha='center', fontsize=12)

    plt.figtext(
        0.5,
        0.01,
        f"Action: Added {water_added}% water to {active_tank_name}",
        ha='center',
        fontsize=12,
        bbox=dict(facecolor='white', edgecolor='black')
    )

    plt.show()

    # Wait before next cycle
    time.sleep(1.5)

print("Simulation Completed Successfully!")
```
